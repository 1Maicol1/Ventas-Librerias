```python
import pandas as pd

# Carga del dataset sin crear columna índice adicional
df_libros = pd.read_csv('ventas_libreria_cultural.csv')

# Visualizamos las primeras filas
df_libros.head()
```

```python
# Extracción de columna como Serie
genero_serie = df_libros['Genero_Libro']

# Nueva tabla (DataFrame) con las tres columnas seleccionadas
df_seleccion = df_libros[['Sucursal', 'Formato', 'Precio_Lista']]

# Verificamos la tabla seleccionada
df_seleccion.head()
```

```python
# Ejercicio 3: Acceso por Etiqueta (.loc[]) para consultar la fila con índice exactamente 310
fila_310 = df_libros.loc[310]
print("--- Ejercicio 3: Fila 310 ---")
print(fila_310)
print("\n")
```

```python
# Ejercicio 4: Tratamiento de Redundancias (.drop_duplicates())
total_antes = len(df_libros)
df_libros = df_libros.drop_duplicates()
total_despues = len(df_libros)
print(f"--- Ejercicio 4: Duplicados ---")
print(f"Filas antes: {total_antes} | Filas después de limpiar: {total_despues}")
print("\n")
```

```python
# Ejercicio 5: Filtrado por Volumen de Compra (Unidades_Vendidas > 3)
compras_mayoristas = df_libros[df_libros['Unidades_Vendidas'] > 3]
print("--- Ejercicio 5: Compras Mayoristas (> 3 unidades) ---")
display(compras_mayoristas.head())
```

```python
# Ejercicio 6: Filtrado Categórico con .isin() (Campus Universitario y Galería de Arte)
sucursales_seleccionadas = ['Campus Universitario', 'Galería de Arte']
df_sucursales = df_libros[df_libros['Sucursal'].isin(sucursales_seleccionadas)]
print("--- Ejercicio 6: Sucursales seleccionadas ---")
display(df_sucursales.head())
print("\n")
```

```python
# Ejercicio 7: Filtrado por Rango de Precios entre $45,000 y $75,000 inclusive
df_precios = df_libros[df_libros['Precio_Lista'].between(45000, 75000)]
print("--- Ejercicio 7: Precios entre 45,000 y 75,000 ---")
display(df_precios.head())
print("\n")
```

```python
# Ejercicio 8: Consultas Directas con .query() (Transferencia QR y Unidades >= 2)
df_query = df_libros.query("Metodo_Pago == 'Transferencia QR' and Unidades_Vendidas >= 2")
print("--- Ejercicio 8: Consulta con .query() ---")
display(df_query.head())
```

```python
# Ejercicio 9: Filtrado Compuesto con Métodos de Texto (Genero_Libro inicia con "Ensayo" y Precio > 50,000)
df_texto = df_libros[df_libros['Genero_Libro'].str.startswith('Ensayo', na=False) & (df_libros['Precio_Lista'] > 50000)]
print("--- Ejercicio 9: Ensayos mayores a $50,000 ---")
display(df_texto.head())
print("\n")
```

```python
# Ejercicio 10: Imputación Estratificada de Faltantes (Precio_Lista con la media, Descuento_Aplicado con 0.0)
df_libros['Precio_Lista'] = df_libros['Precio_Lista'].fillna(df_libros['Precio_Lista'].mean())
df_libros['Descuento_Aplicado'] = df_libros['Descuento_Aplicado'].fillna(0.0)
print("--- Ejercicio 10: Valores nulos imputados correctamente ---")
print(df_libros[['Precio_Lista', 'Descuento_Aplicado']].isnull().sum())
print("\n")
```

```python
# Ejercicio 11: Transformación con .apply() para crear Total_Venta
def calcular_total(fila):
    return (fila['Precio_Lista'] * (1 - fila['Descuento_Aplicado'])) * fila['Unidades_Vendidas']

df_libros['Total_Venta'] = df_libros.apply(calcular_total, axis=1)
print("--- Ejercicio 11: Columna Total_Venta creada ---")
display(df_libros[['ID_Transaccion', 'Precio_Lista', 'Descuento_Aplicado', 'Unidades_Vendidas', 'Total_Venta']].head())
```

```python
# Ejercicio 12: Renombrado exacto y Reordenamiento (ID_Transaccion y Total_Venta al inicio)
df_libros = df_libros.rename(columns={
    'Genero_Libro': 'Categoria',
    'Descuento_Aplicado': 'Descuento'
})

# Reordenar columnas clave al inicio
columnas_inicio = ['ID_Transaccion', 'Total_Venta']
otras_columnas = [col for col in df_libros.columns if col not in columnas_inicio]
df_libros = df_libros[columnas_inicio + otras_columnas]

print("--- Ejercicio 12: Renombrado y Reorganización ---")
display(df_libros.head(2))
print("\n")
```

```python
# Ejercicio 13: Agrupación por Sucursal (Promedio de Precio_Lista y Suma de Total_Venta)
resumen_sucursal = df_libros.groupby('Sucursal').agg({
    'Precio_Lista': 'mean',
    'Total_Venta': 'sum'
}).reset_index()
print("--- Ejercicio 13: Agrupación y Métricas por Sucursal ---")
display(resumen_sucursal)
print("\n")
```

```python
# Ejercicio 14: Función personalizada rango y aplicación con .agg() por Formato
def rango(series):
    return series.max() - series.min()

resumen_rango = df_libros.groupby('Formato').agg({
    'Precio_Lista': rango,
    'Total_Venta': rango
}).reset_index()
print("--- Ejercicio 14: Métricas de Rango Personalizadas por Formato ---")
display(resumen_rango)
print("\n")
```

```python
# Ejercicio 15: Agrupamiento Multinivel (Sucursal y Categoria)
resumen_multinivel = df_libros.groupby(['Sucursal', 'Categoria']).agg({
    'Total_Venta': 'mean',
    'Unidades_Vendidas': 'sum'
}).reset_index()
print("--- Ejercicio 15: Agrupamiento Multinivel ---")
display(resumen_multinivel.head(10))
```

```python
# Exportar dataset final actualizado
df_libros.to_csv('ventas_libreria_cultural_limpio.csv', index=False)
```
